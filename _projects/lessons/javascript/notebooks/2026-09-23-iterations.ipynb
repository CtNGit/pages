## 1. Reference Guide

Every example and hack in this lesson is part of the same video game: you play through **levels**, carry an **inventory**, have **player stats**, lose **lives**, wait to **respawn**, and collect **coins**.

### Key Vocabulary

| Term | Definition | Game example |
| --- | --- | --- |
| **Iteration** | One pass through the body of a loop | Playing one level |
| **Initialization** | Sets the counter once, before the loop starts | `let level = 1` |
| **Condition** | Checked before each pass. `false` ends the loop | `level <= 5` |
| **Increment** | Updates the counter after each pass | `level++` |
| **Infinite loop** | A loop whose condition never becomes false | Lives that never go down |
| **Key** | The name of a property in an object | `health` in `{health: 100}` |

### Loop Grammar

| Loop | When to use it |
| --- | --- |
| `for (let level = 1; level <= 5; level++)` | Repeat a set number of times |
| `for (let i = 0; i < inventory.length; i++)` | Visit every item in an array |
| `for (let stat in player)` | Visit every key of an object |
| `while (lives > 0)` | Repeat until something changes. Set the start before it and change the counter inside it |

Count by 2s with `level += 2`, and count down with `lives--`.

### Tech Talk (5 minutes)

An **iteration** repeats a block of code. Instead of writing `console.log` once for every level, you write it once inside a loop.

* **The Rule:** Every loop needs a start, a condition, and a change.
* Do this: `for (let level = 1; level <= 5; level++) { console.log("Level " + level); }`
* Don't do this: `console.log("Level 1"); console.log("Level 2"); console.log("Level 3"); ...`

One loop handles 5 levels or 500, and you change one number to change how many run.

```javascript
for (let level = 1; level <= 5; level++) {   // start; condition; change
    console.log("Level " + level);
}

let lives = 3;                               // start
while (lives > 0) {                          // condition
    console.log("Lives left: " + lives);
    lives--;                                 // change
}
```

Use `for` when you know how many times to repeat (5 levels), and `while` when you only know the rule for stopping (until you run out of lives).

**Remember:** if the change is missing, the condition never becomes false and you get an **infinite loop**.

## 5. Code Examples

### A. for Loops: Level Select

```javascript
for (initialization; condition; increment) {
    // Code to execute
}
```

* **Initialization** runs once at the start: `let level = 1`.
* **Condition** is checked before every pass: `level <= 5`.
* **Increment** runs after every pass: `level++`.

**Try it:** this level select has a syntax error. Find it, fix it, and press **Run**. Then change it to show only the odd levels with `level += 2`.

````python
%%js
// CODE_RUNNER: Fix the syntax error, run it, then show only the odd levels

for (let level = 1, level <= 5; level++) {
    console.log("Level " + level + " unlocked");
}
````

### B. Arrays and Objects: Inventory and Player Stats

A `for` loop visits every item in an array by its index `i`, and stops after the last one because of `i < inventory.length`.

A `for...in` loop visits every **key** of an object, and `player[stat]` gets its **value**.

**Try it:** add two items to the inventory, add a `coins` stat to the player, and run it again.

````python
%%js
// CODE_RUNNER: Add two items to the inventory and a coins stat to the player, then run it again

let inventory = ["Sword", "Shield", "Health Potion"];

console.log("Inventory:");
for (let i = 0; i < inventory.length; i++) {
    console.log((i + 1) + ". " + inventory[i]);
}

const player = {
    name: "Casey",
    level: 3,
    health: 100,
    speed: 8
};

console.log("Player stats:");
for (let stat in player) { // for...in gives each key; player[stat] gives its value
    console.log(stat + ": " + player[stat]);
}
````

### C. while Loops: Losing Lives

A `while` loop repeats **as long as its condition is true**. It fits when you know the rule for stopping but not how many passes it will take.

A `while` loop always needs:

* **A starting value** before the loop: `let lives = 3`
* **A condition** checked before each pass: `lives > 0`
* **A change** inside the loop that eventually makes the condition false: `lives--`

**Try it:** start with 5 lives, then make each hit cost 2 lives. How many hits until Game Over now?

````python
%%js
// CODE_RUNNER: Start with 5 lives, then make each hit cost 2 lives

let lives = 3;
let hits = 0;

while (lives > 0) {
    hits++;
    lives -= 1;
    console.log("Hit " + hits + "! Lives left: " + lives);
}

console.log("Game Over after " + hits + " hits");
````
